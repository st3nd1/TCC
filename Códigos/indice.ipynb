### Masculino

```python
tabua_referencia = bib.Calculos.calcular_tabua(df_tabuas_masc["IBGE 2024 Extrap M"])

faixas_ativos = [(18,25), (26,30), (31,35), (36,40), (41,45), (46,50), (51,55), (56,60), (61,64)]
faixas_aposentados = [(65,70), (71,75), (76,80), (81,85), (86,90), (91,95), (96, 100)]

lista_proporcoes = list(proporcoes_masc_jovem.values())
proporcoes_ativos     = lista_proporcoes[:9]   # faixas 18–64
proporcoes_aposentados = lista_proporcoes[9:]  # faixas 65–100

# Referência
lx_ativos_referencia = [
    tabua_referencia.loc[(tabua_referencia['Idade'] >= i) & (tabua_referencia['Idade'] <= f), 'lx'].sum()
    for i, f in faixas_ativos
]
lx_aposentados_referencia = [
    tabua_referencia.loc[(tabua_referencia['Idade'] >= i) & (tabua_referencia['Idade'] <= f), 'lx'].sum()
    for i, f in faixas_aposentados
]

sp_ativo_referencia      = sum(v * p for v, p in zip(lx_ativos_referencia,      proporcoes_ativos))
sp_aposentados_referencia = sum(v * p for v, p in zip(lx_aposentados_referencia, proporcoes_aposentados))

lista_resultados = []

for tabua in tabuas_masculinas:
    if tabua != "IBGE 2024 Extrap M":
        tabua_comparativa = bib.Calculos.calcular_tabua(df_tabuas_masc[tabua])

        lx_ativos_comparativa = [
            tabua_comparativa.loc[(tabua_comparativa['Idade'] >= i) & (tabua_comparativa['Idade'] <= f), 'lx'].sum()
            for i, f in faixas_ativos
        ]
        lx_aposentados_comparativa = [
            tabua_comparativa.loc[(tabua_comparativa['Idade'] >= i) & (tabua_comparativa['Idade'] <= f), 'lx'].sum()
            for i, f in faixas_aposentados
        ]

        sp_ativo_comparativa      = sum(v * p for v, p in zip(lx_ativos_comparativa,      proporcoes_ativos))
        sp_aposentados_comparativa = sum(v * p for v, p in zip(lx_aposentados_comparativa, proporcoes_aposentados))

        razao_ativo       = ((sp_ativo_comparativa      / sp_ativo_referencia)      - 1) * 100
        razao_aposentados = ((sp_aposentados_comparativa / sp_aposentados_referencia) - 1) * 100

        lista_resultados.append({
            "tabua": tabua,
            "razao ativo": razao_ativo,
            "razao aposentado": razao_aposentados,
            "ex sup": tabua in lista_tabua_exsup_masc,
            "RM<":   tabua in lista_tabua_pminf_masc
        })

df_ind = pd.DataFrame(lista_resultados)
df_ind[df_ind["RM<"] == True]
```

## Pirâmides

```python
# Confugruação do subplot
fig, axes = plt.subplots(2, 2, figsize=(20, 16))

# Lista de tuplas com as idades simuladas e posições das pirâmides 
configs = [(df_idades_bimodal, "Bimodal", axes[1, 0]), (df_idades_equilibrada, "Equilibrada", axes[0, 1]),
    (df_idades_jovem, "Jovem", axes[0, 0]), (df_idades_idosa, "Idosa", axes[1, 1]),]

# Ordem das faixas etárias
bins_edges = [18, 26, 31, 36, 41, 46, 51, 56, 61, 66, 71, 76, 81, 86, 91, 96, 101]

# Faixas etárias
labels = ["18-25", "26-30", "31-35", "36-40", "41-45", "46-50", "51-55", "56-60", 
    "61-65", "66-70", "71-75", "76-80", "81-85", "86-90", "91-95", "96-100"]

# Loop para cada subplot
for df, titulo, ax in configs:
    
    # Cópia dos dados
    df = df.copy()
    
    # Converter idades inteiras (bins_edge) em grupos (categóricos) de faixas etárias
    df['Faixa Etária'] = pd.cut(df["Idade"], bins=bins_edges, labels=labels, right=False)

    # Agrupando as idades por genero e faixa etária
    pyramid_df = df.groupby(["Faixa Etária", "Gênero"], observed=False).size().unstack().fillna(0)

    # Lado masculino da pirâmide
    ax.barh(pyramid_df.index, -pyramid_df["Masculino"], height=1.0, color="skyblue", edgecolor="black", linewidth=0.5, hatch="oo", label="Masculino")

    # Lado feminino da pirâmide
    ax.barh(pyramid_df.index, pyramid_df["Feminino"], height=1.0, color="pink", edgecolor="black", linewidth=0.5, hatch="xx", label="Feminino")

    # Valor máximo de cada lado da pirâmide
    max_value = max( pyramid_df["Masculino"].max() + 1000, pyramid_df["Feminino"].max() + 1000)
    ax.set_xlim(-max_value, max_value)

    # Eixo central da pirâmide
    ax.axvline(0, color="black", linewidth=1)

    # Formatação do eixo x
    ax.xaxis.set_major_formatter(FuncFormatter(bib.Auxiliares.formatar_milhar_abs))
    ax.set_title(f"População {titulo}", fontsize=22, fontweight="bold")
    ax.set_xlabel("Quantidade de Segurados", fontsize=22)
    ax.set_ylabel("Faixa Etária", fontsize=22)
    ax.tick_params(axis='both', labelsize=22)

    # Legenda
    ax.legend(loc='upper right', frameon=True, framealpha=1, facecolor='white', title='Gênero', title_fontsize=22, fontsize=22)

# Ajuste do tamanho torno dos subplots
plt.tight_layout()

# Estilo
plt.style.use("bmh")

# Salver em pdf
#plt.savefig("piramides_etarias.pdf")

# Mostrar gráfico
plt.show()
```

## Descritiva

```python
# Agrupar os DataFrames em um dicionário
bases_simuladas = {"Jovem": df_idades_jovem,  "Equilibrada": df_idades_equilibrada, "Bimodal": df_idades_bimodal, "Idosa": df_idades_idosa}

# Criar um dicionário para armazenar os resultados
resultados_estatisticos = {}

# Iterar sobre cada base para calcular as métricas
for nome_base, df in bases_simuladas.items():

    # Vetor das idades
    idades = df["Idade"]
    
    # Somar homens <= 64 e mulheres < 62
    soma_jovens = (((df["Gênero"] == "Masculino") & (df["Idade"] <= 64)).sum() + ((df["Gênero"] == "Feminino") & (df["Idade"] < 62)).sum())
    
    # Somar homens >= 65 e mulheres >= 62
    soma_idosos = (((df["Gênero"] == "Masculino") & (df["Idade"] >= 65)).sum() + ((df["Gênero"] == "Feminino") & (df["Idade"] >= 62)).sum())
    
    # Calcular a Maturidade
    if soma_idosos > 0:
        maturidade = soma_jovens / soma_idosos
    else:
        maturidade = np.nan
    
    # Preencher as estatísticas
    resultados_estatisticos[nome_base] = {
        "Idade Média": idades.mean(),
        "Idade Mediana": idades.median(),
        "Idade Mínima": idades.min(),
        "1º Quartil (25%)": idades.quantile(0.25),
        "2º Quartil (50%)": idades.quantile(0.50),
        "3º Quartil (75%)": idades.quantile(0.75),
        "Idade Máxima": idades.max(),
        "Maturidade da Massa": maturidade
    }

# Transformar o dicionário em um DataFrame
df_resumo_estatistico = pd.DataFrame(resultados_estatisticos)

# Arredondar os valores para 2 casas decimais
df_resumo_estatistico = df_resumo_estatistico.round(2)
print(df_resumo_estatistico)
```

# Reservas matemáticas - base equilibrada

```python
# Calcular idade média
idade_media_equilibrada = int(base_cadastral_equilibrada["Idade"].mean())
```

## Masculino

### IEN

```python
# Calculando a RM e testando o Art. 36 para a base cadastral

# Filtrar a base masculina
base_masc_equilibrada = base_cadastral_equilibrada[base_cadastral_equilibrada["Gênero"] == "Masculino"]

# Separar bases para PMBaC e PMBC
base_PMBaC = base_masc_equilibrada[base_masc_equilibrada["Idade"] < 65]
base_PMBC = base_masc_equilibrada[base_masc_equilibrada["Idade"] >= 65]

# Listas locais para armazenar resultados
lista_vabf, lista_vacf, lista_ex = [], [], []
lista_PMBAC, lista_PMBC, lista_RM = [], [], []

# Loop para iterar sobre as tábuas e calcular a reserva matemática
for tabua in tabuas_masculinas:

    # Probabilidade de morte da tábua
    vetor_qx = df_tabuas_masc[tabua]

    # Calcular a tábua de vida
    tabua_de_vida = bib.Calculos.calcular_tabua(vetor_qx)

    # Calcular PMBaC
    vabf_pmbac = bib.Calculos.VABF_PMBaC(base_PMBaC, tabua_de_vida, 0.05, 0.01)
    vacf_pmbac = bib.Calculos.VACF_PMBaC_INE(base_PMBaC, tabua_de_vida, vabf_pmbac, 0.05)

    # PMBaC individual (cada ativo)
    PMBaC_individual = vabf_pmbac - vacf_pmbac

    # PMBaC total (soma de todos os ativos)
    PMBaC_total = np.sum(PMBaC_individual)

    # Calcular PMBC
    PMBC_individual = bib.Calculos.VABF_PMBC(base_PMBC, tabua_de_vida, 0.05, 0.01)

    # PMBC total (soma de todos os aposentados)
    PMBC_total = np.sum(PMBC_individual)

    # Cálculo RM TOTAL
    RM_total = PMBaC_total + PMBC_total

    # Obtém a expectativa de vida na idade média
    ex_im = tabua_de_vida.loc[idade_media_equilibrada,"ex"]

    # Adiciona os valores as suas devidas listas
    lista_vabf.append(np.sum(vabf_pmbac))
    lista_vacf.append(np.sum(vacf_pmbac))
    lista_PMBAC.append(PMBaC_total)
    lista_PMBC.append(PMBC_total)
    lista_RM.append(RM_total)
    lista_ex.append(round(ex_im, 2))

# DataFrame com os resultados
df_comp = pd.DataFrame({"Tábua": tabuas_masculinas, "VABF": lista_vabf, "VACF": lista_vacf,
                        "PMBaC": lista_PMBAC, "PMBC": lista_PMBC, "RM": lista_RM,"ex": lista_ex})

# Comparar com a primeira linha
ref_ex = df_comp.loc[0, "ex"]
ref_RM = df_comp.loc[0, "RM"]

# Verificação do Art 36
df_comp["Art36_OK"] = (df_comp["ex"] >= ref_ex) & (df_comp["RM"] < ref_RM)


# Aplica o formato monetário brasileiro
for col in ["VABF", "VACF", "PMBaC", "PMBC", "RM"]:
    df_comp[col] = df_comp[col].apply(bib.Auxiliares.formatar_brasileiro)

#df_comp
# Concatenar primeira linha com linhas que atendem Art. 36
df_resultado = pd.concat([df_comp.head(1), df_comp[df_comp["Art36_OK"] == True]]).drop_duplicates()

display(df_resultado)
```
